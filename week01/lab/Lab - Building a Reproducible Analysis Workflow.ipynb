```python
# Sort by mean from largest to smallest so the highest-selling genre per game is on top.
vgsales_df.groupby("Genre")["Global_Sales"].agg(["count", "mean"]).sort_values("mean", ascending=False).round(3)
```

```python
# Sort the same summary by count so the genre with the most games is on top.
vgsales_df.groupby("Genre")["Global_Sales"].agg(["count", "mean"]).sort_values("count", ascending=False).round(3)
```

What percentage of games in the dataset sold at least 1 million copies worldwide, and how does that percentage compare between games published by Nintendo and games from all other publishers? ### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer:

For the overall percentage I use all 16,598 rows and `Global_Sales`; the denominator is 16,598 and the numerator is the number of rows with `Global_Sales` >= 1.00 million. For the comparison I use `Publisher` and `Global_Sales` and drop the 58 rows whose publisher is missing, because they cannot be assigned to either group. That leaves 16,540 rows: the Nintendo group (publisher exactly "Nintendo") and the all-other-publishers group. Each group's percentage uses its own row count as the denominator. I create two new features: `million_seller` (True when `Global_Sales` >= 1) and `publisher_group` ("Nintendo" or "All other publishers").


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# New feature: True when the game sold at least 1 million copies worldwide.
vgsales_df["million_seller"] = vgsales_df["Global_Sales"] >= 1

# Overall: the mean of a True/False feature is the proportion of True values.
overall_pct = vgsales_df["million_seller"].mean() * 100
print("Million-sellers:", vgsales_df["million_seller"].sum(), "of", len(vgsales_df), "games")
print(f"Overall percentage: {overall_pct:.2f}%")
```

```python
# Drop the 58 rows with no publisher, then label each remaining game Nintendo or not.
with_publisher_df = vgsales_df.dropna(subset=["Publisher"]).copy()
with_publisher_df["publisher_group"] = with_publisher_df["Publisher"].where(
    with_publisher_df["Publisher"] == "Nintendo", "All other publishers"
)
with_publisher_df["publisher_group"].value_counts()
```

```python
# Percentage of million-sellers in each publisher group.
# count = games in the group (the denominator), sum = million-sellers, mean = proportion.
group_summary = with_publisher_df.groupby("publisher_group")["million_seller"].agg(["count", "sum", "mean"])
group_summary["percent"] = (group_summary["mean"] * 100).round(1)
group_summary
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:

Overall, 2,081 of the 16,598 games (12.54%) sold at least 1 million copies. Among games with a recorded publisher, 339 of Nintendo's 703 games are million-sellers, 48.2%, compared with 1,739 of the 15,837 games from all other publishers, 11.0%. So a Nintendo-published game in this dataset is about 4.4 times as likely to be a million-seller. The `million_seller` feature turns the sales threshold into True/False, the mean of that feature is the proportion of True values, and grouping by `publisher_group` gives that proportion separately for each group with its own denominator (`count`).

## Investigation 3

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 3?**

Your answer:

Do handheld platforms draw a larger share of their sales from Japan than home consoles and PC do? "Handheld" means the platform codes GB, GBA, DS, 3DS, PSP, PSV, GG, and WS; every other platform code (Wii, PS2, X360, PC, NES, and so on) is "Home console or PC". "Share from Japan" is the total `JP_Sales` of all games in the group divided by the total `Global_Sales` of the same games, times 100, so it is a percentage of copies sold.


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer:

I predict handhelds will have a clearly larger Japan share, something like 25% versus about 10% for home consoles and PC. Handheld gaming has historically been more popular in Japan (Pokémon, Monster Hunter, and the DS and PSP libraries), while Xbox and PC sales come almost entirely from North America and Europe.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer:

I use all 16,598 rows and three features: `Platform` (to build the group), `JP_Sales`, and `Global_Sales`, all complete with 0 missing values. I create a new categorical feature `platform_type` that is "Handheld" when `Platform` is in the handheld list and "Home console or PC" otherwise. The percentage for each group is sum of `JP_Sales` divided by sum of `Global_Sales` within the group, so the denominator is the group's total worldwide sales in millions of copies. I sum first instead of averaging each game's own Japan share, so a game that sold 30 million copies counts 30 million times more than a game that sold 0.1 million. No rows are filtered out.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
# New feature: group each platform code as Handheld or Home console or PC.
handheld_platforms = ["GB", "GBA", "DS", "3DS", "PSP", "PSV", "GG", "WS"]
vgsales_df["platform_type"] = vgsales_df["Platform"].where(
    ~vgsales_df["Platform"].isin(handheld_platforms), "Handheld"
)
vgsales_df.loc[vgsales_df["platform_type"] != "Handheld", "platform_type"] = "Home console or PC"
vgsales_df["platform_type"].value_counts()
```

```python
# Total Japan sales and total global sales in each group, then Japan's share as a percentage.
region_summary = vgsales_df.groupby("platform_type")[["JP_Sales", "Global_Sales"]].sum()
region_summary["japan_share_percent"] = (region_summary["JP_Sales"] / region_summary["Global_Sales"] * 100).round(1)
region_summary
```

```python
# The same calculation for each individual handheld platform, largest total sales first.
handheld_detail = vgsales_df[vgsales_df["platform_type"] == "Handheld"].groupby("Platform")[["JP_Sales", "Global_Sales"]].sum()
handheld_detail["japan_share_percent"] = (handheld_detail["JP_Sales"] / handheld_detail["Global_Sales"] * 100).round(1)
handheld_detail.sort_values("Global_Sales", ascending=False)
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer:

Handheld platforms drew 25.2% of their sales from Japan (504.58 of 2,003.57 million copies) versus 11.4% for home consoles and PC (786.44 of 6,916.87 million), so the handheld share is a little more than double. The groupby sums `JP_Sales` and `Global_Sales` across every game in each group, and dividing the two sums gives the fraction of all copies in that group that were sold in Japan. The per-platform table shows the pattern holds inside the handheld group: the 3DS is 39.3% Japan, the Game Boy 33.3%, the PSP 25.9%, and the DS 21.3%, with only the GBA (14.9%) coming close to the home-console figure.


### Step 4: Check

Perform a sanity check tied to the meaning of your question. Explain what values or patterns would make you suspicious and whether your result makes sense. Depending on the analysis, consider plausible values and units, whether indicators match original labels, or whether the ranges within comparison groups match their definitions. Use additional code only when it helps this check; do not simply repeat the initial row count or missing-value inspection.
```python
# Check 1: every platform code landed in the intended group.
for group, codes in vgsales_df.groupby("platform_type")["Platform"].unique().items():
    print(f"{group}: {sorted(codes)}")

# Check 2: the four regional sales features should add up to Global_Sales (allowing for rounding to 0.01).
region_gap = (vgsales_df[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].sum(axis=1) - vgsales_df["Global_Sales"]).abs()
print()
print("Largest gap between regional sum and Global_Sales:", round(region_gap.max(), 2), "million copies")

# Check 3: Japan's share for the whole dataset, which should sit between the two group values.
print("Overall Japan share:", round(vgsales_df["JP_Sales"].sum() / vgsales_df["Global_Sales"].sum() * 100, 1), "%")
```

**What would make you suspicious, and why? Does your evidence make sense in context? Describe any issue you found and how you addressed it.**

Your answer:

I would be suspicious if a Japan share came out above 100% or below 0%, if the regional sales did not add up to global sales (which would mean `JP_Sales` and `Global_Sales` are not measuring the same copies), or if a platform ended up in the wrong group. The checks pass: the Handheld group contains exactly GB, DS, GBA, 3DS, PSP, PSV, WS, and GG and nothing else; the largest gap between the four regions and `Global_Sales` is 0.02 million, which is just rounding to two decimals; and the overall Japan share is 14.5%, which sits between the 11.4% and 25.2% group values as it must. Two tiny platforms, WS and GG, show a 100% Japan share, but they total only 1.46 million copies between them, so they barely move the handheld figure. The result also makes sense in context, since the PS (19.1%) is the only major home console above 11%, while the X360 (1.3%) and PC (0.1%) are almost entirely non-Japanese.

Yes. Japan accounts for 25.2% of all copies sold on handheld platforms in this dataset, compared with 11.4% on home consoles and PC, so the handheld share is about 2.2 times larger. This matches my prediction almost exactly (I guessed about 25% versus about 10%).
